# Exercises: Products and Quotients of Vector Spaces
*Axler, **Linear Algebra Done Right**, 4th ed., §3E*

## Exercise 3

> Suppose $V_1, \ldots, V_m$ are vector spaces. Prove that $\mathcal{L}(V_1 \times \cdots \times V_m, W)$ and $\mathcal{L}(V_1, W) \times \cdots \times \mathcal{L}(V_m, W)$ are isomorphic vector spaces.

**Idea.** A linear map out of a product is determined by what it does on each factor separately. So we restrict $T$ to each slot, and in the other direction we add the pieces back together.

**Setup.** For $k = 1, \ldots, m$ define the *inclusion* $R_k \colon V_k \to V_1 \times \cdots \times V_m$ by
$$R_k(v) = (0, \ldots, 0, \underbrace{v}_{k\text{th slot}}, 0, \ldots, 0).$$
Each $R_k$ is linear, since addition and scalar multiplication in the product are done one slot at a time. Note also that every element of the product splits as
$$(v_1, \ldots, v_m) = R_1 v_1 + \cdots + R_m v_m. \tag{$*$}$$

**Proof.** Define
$$\Phi \colon \mathcal{L}(V_1 \times \cdots \times V_m, W) \to \mathcal{L}(V_1, W) \times \cdots \times \mathcal{L}(V_m, W), \qquad \Phi(T) = (T R_1, \ldots, T R_m).$$
Each $T R_k$ is a composition of linear maps, so $T R_k \in \mathcal{L}(V_k, W)$ and $\Phi$ is well defined. We show that $\Phi$ is an isomorphism.

*$\Phi$ is linear.* For $S, T \in \mathcal{L}(V_1 \times \cdots \times V_m, W)$ and $\lambda \in \mathbf{F}$, composition distributes over addition and scalar multiplication (3.8), so
$$\Phi(S + T) = \big((S+T)R_1, \ldots, (S+T)R_m\big) = (SR_1 + TR_1, \ldots, SR_m + TR_m) = \Phi(S) + \Phi(T),$$
$$\Phi(\lambda T) = \big((\lambda T)R_1, \ldots, (\lambda T)R_m\big) = \lambda\,(TR_1, \ldots, TR_m) = \lambda\,\Phi(T).$$

*$\Phi$ is injective.* Suppose $\Phi(T) = 0$, so $T R_k = 0$ for every $k$. Then for any $(v_1, \ldots, v_m)$, by $(*)$,
$$T(v_1, \ldots, v_m) = T(R_1 v_1) + \cdots + T(R_m v_m) = 0 + \cdots + 0 = 0.$$
So $T = 0$, which means $\operatorname{null}\Phi = \{0\}$ and $\Phi$ is injective (3.15).

*$\Phi$ is surjective.* Let $(S_1, \ldots, S_m) \in \mathcal{L}(V_1, W) \times \cdots \times \mathcal{L}(V_m, W)$. Define $T \colon V_1 \times \cdots \times V_m \to W$ by
$$T(v_1, \ldots, v_m) = S_1 v_1 + \cdots + S_m v_m.$$
$T$ is linear, because
$$\begin{aligned}
T\big((v_1, \ldots, v_m) + \lambda(u_1, \ldots, u_m)\big)
&= T(v_1 + \lambda u_1, \ldots, v_m + \lambda u_m) \\
&= \textstyle\sum_{k} S_k(v_k + \lambda u_k) \\
&= \textstyle\sum_{k} S_k v_k + \lambda \sum_{k} S_k u_k \\
&= T(v_1, \ldots, v_m) + \lambda\, T(u_1, \ldots, u_m).
\end{aligned}$$
Also, for $v \in V_k$, every slot of $R_k v$ is $0$ except slot $k$, so $T R_k v = S_k v$. Hence $T R_k = S_k$ for each $k$, and $\Phi(T) = (S_1, \ldots, S_m)$.

Since $\Phi$ is linear, injective, and surjective, it is an isomorphism. So $\mathcal{L}(V_1 \times \cdots \times V_m, W) \cong \mathcal{L}(V_1, W) \times \cdots \times \mathcal{L}(V_m, W)$. $\blacksquare$

**Remarks.**

1. The proof never uses finite dimension, so the result holds for all vector spaces. That is why we construct an explicit isomorphism instead of comparing dimensions.


## Exercise 4

> Suppose $W_1, \ldots, W_m$ are vector spaces. Prove that $\mathcal{L}(V, W_1 \times \cdots \times W_m)$ and $\mathcal{L}(V, W_1) \times \cdots \times \mathcal{L}(V, W_m)$ are isomorphic vector spaces.

**Idea.** This is the mirror image of Exercise 3. A linear map *into* a product is determined by its coordinate maps. So we read off each coordinate of $T$, and in the other direction we stack the pieces side by side.

**Setup.** For $k = 1, \ldots, m$ define the *projection* $P_k \colon W_1 \times \cdots \times W_m \to W_k$ by
$$P_k(w_1, \ldots, w_m) = w_k.$$
Each $P_k$ is linear, since addition and scalar multiplication in the product are done one slot at a time. Note also that every element of the product is determined by its coordinates:
$$w = (P_1 w, \ldots, P_m w) \quad \text{for all } w \in W_1 \times \cdots \times W_m. \tag{$**$}$$

**Proof.** Define
$$\Phi \colon \mathcal{L}(V, W_1 \times \cdots \times W_m) \to \mathcal{L}(V, W_1) \times \cdots \times \mathcal{L}(V, W_m), \qquad \Phi(T) = (P_1 T, \ldots, P_m T).$$
Each $P_k T$ is a composition of linear maps $V \xrightarrow{T} W_1 \times \cdots \times W_m \xrightarrow{P_k} W_k$, so $P_k T \in \mathcal{L}(V, W_k)$ and $\Phi$ is well defined. We show that $\Phi$ is an isomorphism.

*$\Phi$ is linear.* For $S, T \in \mathcal{L}(V, W_1 \times \cdots \times W_m)$ and $\lambda \in \mathbf{F}$, composition distributes over addition and scalar multiplication (3.8), so
$$\Phi(S + T) = \big(P_1(S+T), \ldots, P_m(S+T)\big) = (P_1S + P_1T, \ldots, P_mS + P_mT) = \Phi(S) + \Phi(T),$$
$$\Phi(\lambda T) = \big(P_1(\lambda T), \ldots, P_m(\lambda T)\big) = \lambda\,(P_1T, \ldots, P_mT) = \lambda\,\Phi(T).$$

*$\Phi$ is injective.* Suppose $\Phi(T) = 0$, so $P_k T = 0$ for every $k$. Then for any $v \in V$, by $(**)$,
$$Tv = (P_1 Tv, \ldots, P_m Tv) = (0, \ldots, 0) = 0.$$
So $T = 0$, which means $\operatorname{null}\Phi = \{0\}$ and $\Phi$ is injective (3.15).

*$\Phi$ is surjective.* Let $(S_1, \ldots, S_m) \in \mathcal{L}(V, W_1) \times \cdots \times \mathcal{L}(V, W_m)$. Define $T \colon V \to W_1 \times \cdots \times W_m$ by
$$Tv = (S_1 v, \ldots, S_m v).$$
$T$ is linear, because for $u, v \in V$ and $\lambda \in \mathbf{F}$,
$$\begin{aligned}
T(v + \lambda u)
&= \big(S_1(v + \lambda u), \ldots, S_m(v + \lambda u)\big) \\
&= (S_1 v + \lambda S_1 u, \ldots, S_m v + \lambda S_m u) \\
&= (S_1 v, \ldots, S_m v) + \lambda\,(S_1 u, \ldots, S_m u) \\
&= Tv + \lambda\, Tu.
\end{aligned}$$
Also $P_k T v = S_k v$ for every $v \in V$, so $P_k T = S_k$ for each $k$, and $\Phi(T) = (S_1, \ldots, S_m)$.

Since $\Phi$ is linear, injective, and surjective, it is an isomorphism. So $\mathcal{L}(V, W_1 \times \cdots \times W_m) \cong \mathcal{L}(V, W_1) \times \cdots \times \mathcal{L}(V, W_m)$. $\blacksquare$

**Remarks.**

1. As in Exercise 3, nothing here uses finite dimension, so the result holds for all vector spaces.
2. Compare the two exercises: maps *out of* a product are handled with the inclusions $R_k$ (precompose, $T \mapsto TR_k$), while maps *into* a product are handled with the projections $P_k$ (postcompose, $T \mapsto P_kT$).


## Exercise 5

> For $m$ a positive integer, define $V^m$ by $V^m = \underbrace{V \times \cdots \times V}_{m \text{ times}}$. Prove that $V^m$ and $\mathcal{L}(\mathbf{F}^m, V)$ are isomorphic vector spaces.

**Idea.** A linear map out of $\mathbf{F}^m$ is determined by where it sends the standard basis $e_1, \ldots, e_m$ (3.4). So a list $(v_1, \ldots, v_m)$ of $m$ vectors in $V$ carries exactly the same information as a linear map $\mathbf{F}^m \to V$: send $e_k$ to $v_k$.

**Setup.** Let $e_1, \ldots, e_m$ be the standard basis of $\mathbf{F}^m$, so every $a \in \mathbf{F}^m$ can be written as
$$a = (a_1, \ldots, a_m) = a_1 e_1 + \cdots + a_m e_m. \tag{$\dagger$}$$

**Proof.** Define
$$\Phi \colon V^m \to \mathcal{L}(\mathbf{F}^m, V), \qquad \Phi(v_1, \ldots, v_m)(a_1, \ldots, a_m) = a_1 v_1 + \cdots + a_m v_m.$$
For fixed $(v_1, \ldots, v_m)$, the map $\Phi(v_1, \ldots, v_m)$ is linear, because for $a, b \in \mathbf{F}^m$ and $\lambda \in \mathbf{F}$,
$$\Phi(v_1, \ldots, v_m)(a + \lambda b) = \textstyle\sum_k (a_k + \lambda b_k) v_k = \sum_k a_k v_k + \lambda \sum_k b_k v_k = \Phi(v_1, \ldots, v_m)(a) + \lambda\,\Phi(v_1, \ldots, v_m)(b).$$
So $\Phi(v_1, \ldots, v_m) \in \mathcal{L}(\mathbf{F}^m, V)$ and $\Phi$ is well defined. Note also that
$$\Phi(v_1, \ldots, v_m)(e_k) = v_k \quad \text{for each } k. \tag{$\ddagger$}$$
We show that $\Phi$ is an isomorphism.

*$\Phi$ is linear.* Let $(v_1, \ldots, v_m), (u_1, \ldots, u_m) \in V^m$ and $\lambda \in \mathbf{F}$. For every $a \in \mathbf{F}^m$,
$$\begin{aligned}
\Phi\big((v_1, \ldots, v_m) + (u_1, \ldots, u_m)\big)(a)
&= \Phi(v_1 + u_1, \ldots, v_m + u_m)(a) \\
&= \textstyle\sum_k a_k (v_k + u_k) \\
&= \textstyle\sum_k a_k v_k + \sum_k a_k u_k \\
&= \Phi(v_1, \ldots, v_m)(a) + \Phi(u_1, \ldots, u_m)(a),
\end{aligned}$$
$$\Phi\big(\lambda (v_1, \ldots, v_m)\big)(a) = \Phi(\lambda v_1, \ldots, \lambda v_m)(a) = \textstyle\sum_k a_k (\lambda v_k) = \lambda \sum_k a_k v_k = \lambda\,\Phi(v_1, \ldots, v_m)(a).$$
Since these hold for every $a$, we get $\Phi\big((v_1, \ldots, v_m) + (u_1, \ldots, u_m)\big) = \Phi(v_1, \ldots, v_m) + \Phi(u_1, \ldots, u_m)$ and $\Phi\big(\lambda(v_1, \ldots, v_m)\big) = \lambda\,\Phi(v_1, \ldots, v_m)$.

*$\Phi$ is injective.* Suppose $\Phi(v_1, \ldots, v_m) = 0$, the zero map. Then by $(\ddagger)$,
$$v_k = \Phi(v_1, \ldots, v_m)(e_k) = 0 \quad \text{for each } k.$$
So $(v_1, \ldots, v_m) = (0, \ldots, 0)$, which means $\operatorname{null}\Phi = \{0\}$ and $\Phi$ is injective (3.15).

*$\Phi$ is surjective.* Let $T \in \mathcal{L}(\mathbf{F}^m, V)$. Set $v_k = T e_k$ for $k = 1, \ldots, m$. Then for every $a \in \mathbf{F}^m$, using $(\dagger)$ and the linearity of $T$,
$$\Phi(v_1, \ldots, v_m)(a) = \textstyle\sum_k a_k\, T e_k = T\big(\sum_k a_k e_k\big) = T a.$$
So $\Phi(Te_1, \ldots, Te_m) = T$.

Since $\Phi$ is linear, injective, and surjective, it is an isomorphism. So $V^m \cong \mathcal{L}(\mathbf{F}^m, V)$. $\blacksquare$

**Remarks.**

1. The inverse is explicit: $\Phi^{-1}(T) = (Te_1, \ldots, Te_m)$. It just evaluates $T$ on the standard basis.
2. Nothing here uses finite dimension of $V$, so the result holds for every vector space $V$.
3. This is also a special case of Exercise 3. Take $V_1 = \cdots = V_m = \mathbf{F}$ and $W = V$. Then $\mathcal{L}(\mathbf{F}^m, V) \cong \mathcal{L}(\mathbf{F}, V)^m$. Also $\mathcal{L}(\mathbf{F}, V) \cong V$ via $S \mapsto S(1)$, so $\mathcal{L}(\mathbf{F}, V)^m \cong V^m$.


## Exercise 6

> Suppose $v, x$ are vectors in $V$ and $U, W$ are subspaces of $V$ such that $v + U = x + W$. Prove that $U = W$.

**Idea.** A translate $v + U$ remembers the subspace $U$ it came from: it is the set of all differences of its elements. So two equal translates must come from the same subspace. The plan is to first move both translates to the same base point $v$, and then cancel $v$.

**Setup.** Recall 3.101: for a subspace $W$ and $v, x \in V$,
$$v - x \in W \iff v + W = x + W. \tag{$\S$}$$

**Proof.** Suppose
$$v + U = x + W. \tag{1}$$

*Step 1: $v - x \in W$.* Since $0 \in U$, we have $v = v + 0 \in v + U$. By (1), $v \in x + W$, so $v = x + w_0$ for some $w_0 \in W$. Hence $v - x = w_0 \in W$.

*Step 2: Move to the same base point.* By Step 1 and $(\S)$, $x + W = v + W$. Combining with (1),
$$v + U = v + W. \tag{2}$$

*Step 3: Cancel $v$.* Let $u \in U$. Then $v + u \in v + U$, so by (2), $v + u = v + w$ for some $w \in W$. Subtracting $v$ gives $u = w \in W$. So $U \subseteq W$.

In the same way, let $w \in W$. Then $v + w \in v + W$, so by (2), $v + w = v + u$ for some $u \in U$. So $w = u \in U$, which gives $W \subseteq U$.

Therefore $U = W$. $\blacksquare$

**Remarks.**

1. By symmetry, Step 1 also gives $x - v \in U$. So the hypothesis forces $v - x \in U \cap W$, and since $U = W$, we get $v + U = x + U$ as well.
2. A translate determines its subspace intrinsically: $U = \{\, a - b : a, b \in v + U \,\}$. Applying this to both sides of (1) gives another one-line proof.
